```python
import pandas as pd

file_path = 'enrolment.csv'
df = pd.read_csv(file_path)
print(df.columns)
# print(df.head(10))

print(df['Country'].unique())
print(df['COUNTRY'].unique())
```

```python
# df = df[df['COUNTRY'] == 'AUS']
# df = df[df['Education level'] == 'Upper secondary vocational education']

df = df[df['Year'] == 2020]
df = df[df['Gender'] != 'Total']

df = df[['Field', 'Year', 'Value']]
df['Value'] = df['Value'].fillna(0)

print(df.head(10))
```

```python
df = df[['Field', 'Value']].groupby('Field').sum()
df['categories'] = ['biology chemistry', 'arts literature music psychology', 'business literature', '', 'mathematics physics', '', '', 'biology chemistry psychology', 'computer_science physics', 'biology chemistry physics mathematics', 'computer_science mathematics physics', 'business psychology', 'literature psychology', '']
values = df['Value'].tolist()
df['percentages'] = [100 * v / values[-1] for v in values]
df['Value'] = df['Value'].map('{:.0f}'.format)

df = df[df.index != 'Field unspecified (Broad field level)']
df = df[df.index != 'Generic programmes and qualifications']

print(df)
```

```python
data = {
    'categories': df['categories'].tolist(),
    'value': df['Value'].tolist(),
    'percentages': df['percentages'].tolist()
}
df = pd.DataFrame(data)
print(df)
print()
print(f'--total values: {sum([int(v) for v in df['value'].tolist()[:-1]])}')
print(f'--total percentages: {round(sum(df['percentages'].tolist()[:-1]), 2)}--')


file_path = 'enrolment_by_field_total.csv'
df.to_csv(file_path)
```
