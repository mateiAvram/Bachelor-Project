```python
import pandas as pd
from sql_data_service import SqlDataService
```

```python
data_service = SqlDataService()
```

```python
results = data_service.get_profiling_results()
bots = []
for r in results:
    data = {
        'id': r[0],
        'username': r[1],
        'bot_vector': r[2],
        'user_vector': r[3]
    }
    bots.append(data)

df = pd.DataFrame(bots)
df.to_csv('profiling_multiple.csv', index=False)
df.head(10)
```

```python
results = data_service.get_recommendation_results()
actions = []
for r in results:
    data = {
        'user': r[0],
        'vector': r[1],
        'video_id': r[2],
        'categories': r[3],
        'action_type': r[4],
        'action_duration': r[5],
        'video_duration': r[6]
    }
    actions.append(data)

df = pd.DataFrame(actions)
df.to_csv('recommending_multiple.csv', index=False)
df.head(10)
```

```python
results = data_service.get_training_set()
videos = []
for r in results:
    data = {
        'id': r[0],
        'title': r[1],
        'category': r[2],
        'duration': r[3],
        'tags': r[4],
        'publish_date': r[5],
        'vector': r[6]
    }
    videos.append(data)

df = pd.DataFrame(videos)
df.to_csv('videos_multiple.csv', index=False)
df.head(10)
```
